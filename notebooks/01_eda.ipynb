# Exploratory Data Analysis — Synthetic AMR Demo

Quick look at the generated data before training.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA = Path("../data/processed")
feats = pd.read_csv(DATA / "features.csv")
labels = pd.read_csv(DATA / "labels.csv")
df = feats.merge(labels, on="sample_id")

gene_cols = [c for c in feats.columns if c != "sample_id"]
abx_cols = [c for c in labels.columns if c != "sample_id"]

print(f"Genomes: {len(df)}")
print(f"Genes: {len(gene_cols)}")
print(f"Antibiotics: {len(abx_cols)} — {abx_cols}")
print()
print(df[abx_cols].apply(lambda x: (x=="R").mean()).rename("R_rate"))
```

```python
# Gene prevalence
prev = df[gene_cols].mean().sort_values(ascending=False)
plt.figure(figsize=(10, 5))
prev.plot(kind="bar", color="#4c78a8")
plt.title("Gene carriage prevalence")
plt.ylabel("Fraction of genomes")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Co-occurrence of top 15 genes
top15 = prev.head(15).index.tolist()
corr = df[top15].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, cmap="RdBu_r", center=0, vmin=-1, vmax=1, annot=False)
plt.title("Co-occurrence (Pearson) of top 15 genes")
plt.tight_layout()
plt.show()
```

```python
# Resistance phenotype co-occurrence
phen = df[abx_cols].applymap(lambda x: 1 if x=="R" else 0)
plt.figure(figsize=(6, 4.5))
sns.heatmap(phen.corr(), annot=True, fmt=".2f", cmap="RdBu_r", center=0)
plt.title("Resistance phenotype co-occurrence")
plt.tight_layout()
plt.show()
```
